```python
import requests
import pandas as pd

# Function to get counties in a given state
def get_counties(state):
    data_url = f"https://api.census.gov/data/2023/acs/acs5/profile?get=group(DP05)&ucgid=pseudo(0400000US{state}$0500000)"
    response = requests.get(data_url)
    if response.status_code == 200:
        data = response.json()
        df_data = pd.DataFrame(data[1:], columns=data[0])
        return df_data[["NAME", "GEO_ID"]]
    else:
        print(f"Error fetching counties for state {state}: {response.status_code}")
        return pd.DataFrame()  # Return empty DataFrame in case of error

# Function to get zip codes in a given state and county
def get_zip_codes(state, county_code):
    data_url = f"https://api.census.gov/data/2023/acs/acs5/profile?get=group(DP05)&ucgid=pseudo({county_code}$8600000)"
    response = requests.get(data_url)
    if response.status_code == 200:
        data = response.json()
        df_data = pd.DataFrame(data[1:], columns=data[0])
        return df_data
    else:
        print(f"Error fetching zip codes for county {county_code} in state {state}: {response.status_code}")
        return pd.DataFrame()  # Return empty DataFrame in case of error

# Function to aggregate data for all zip codes in all counties of a state
def collect_zipcode_data(state):
    # Get all counties in the state
    counties_df = get_counties(state)
    if counties_df.empty:
        return pd.DataFrame()  # Return empty if no counties were found
    
    final_df = pd.DataFrame()  # DataFrame to store the final result

    # Iterate over each county in the state
    for index, row in counties_df.iterrows():
        county = row['NAME']  # Extract county ID from the row (adjust based on your actual data structure)
        county_code = row['GEO_ID']  # Extract county code from the row (adjust based on your actual data structure)
        print(f"Fetching data for county: {county_code} - {county}")
        
        # Get zip codes for the county
        zip_codes_df = get_zip_codes(state, county_code)
        if not zip_codes_df.empty:
            # Add county and state information to the zip code data
            zip_codes_df['County'] = county
            zip_codes_df['County Code'] = county_code
            zip_codes_df['State'] = state
            final_df = pd.concat([final_df, zip_codes_df], ignore_index=True)
    
    return final_df

# Example: Collect data for a given state (e.g., state '37' for North Carolina)
state = '37'  # Replace with the state FIPS code you want to query
zipcode_data = collect_zipcode_data(state)

# Display or save the collected data
print(zipcode_data.head())
#zipcode_data.to_csv("zipcode_data.csv", index=False)  # Optionally save to CSV
```

```python
import re

columns = df.columns.to_list()
df.columns
```

```python
# Classify gender-related columns
gender_male = []
gender_female = []
gender_ratio = []

for col in columns:
    col_lower = col.lower()
    # Direct Male/Female columns
    if 'ratio' in col_lower:
        gender_male.append(col)
    elif 'female' in col_lower:
        gender_female.append(col)
    # Gender Ratio columns
    elif 'male' in col_lower:
        gender_ratio.append(col)

print("Male Columns:", gender_male)
print("Female Columns:", gender_female)
print("Gender Ratio Columns:", gender_ratio)
```

```python
# Classify age-related columns
age_brackets = []
age_groups = []
age_overall = []
age_citizen = []

for col in columns:
    # Rule: Age Bracket (has "X to Y years" pattern)
    if 'Citizen' in col:
        age_citizen.append(col)
    elif re.search(r'\d+\s*to\s*\d+\s*years', col):
        age_brackets.append(col)
    # Rule: Age Group (contains "under", "and over", or specific keywords)
    elif any(keyword in col.lower() for keyword in ['under', 'and over']):
        age_groups.append(col)
    elif 'Median age' in col:
        age_overall.append(col)

print("Age Brackets:", age_brackets)
print("Age Groups:", age_groups)
print("Overall Age:", age_overall)
print("Citizens Age:", age_citizen)
```

```python
# Race and ethnicity categories
single_race = []
asian_subcategories = []
mixed_race = []
native_american_indigenous = []
hispanic_ethnicity = []
general_race = []

# Mapping logic
for col in columns:
    col_lower = col.lower()
    
    # Standard single race
    if any(race in col_lower for race in ['white', 'black or african american', 'native hawaiian']):
        single_race.append(col)

    # Asian subcategories
    elif any(asian_group in col_lower for asian_group in ['asian indian', 'chinese', 'filipino', 'japanese', 'korean', 'vietnamese', 'other asian']):
        asian_subcategories.append(col)
    
    # Mixed race categories
    elif 'and' in col_lower or 'two races' in col_lower or 'three' in col_lower:
        mixed_race.append(col)
    
    # Native American or Indigenous groups
    elif any(group in col_lower for group in ['american indian', 'alaska native', 'aztec', 'navajo', 'blackfeet', 'inupiat', 'maya']):
        native_american_indigenous.append(col)
    
    # Hispanic or Latino categories
    elif any(term in col_lower for term in ['hispanic', 'latino', 'mexican', 'puerto rican', 'cuban']):
        hispanic_ethnicity.append(col)
    
    # General categories
    else:
        general_race.append(col)

# Display categorized results
print("Single Race Categories:", single_race)
print("Asian Subcategories:", asian_subcategories)
print("Mixed Race Categories:", mixed_race)
print("Native American/Indigenous Groups:", native_american_indigenous)
print("Hispanic/Latino Categories:", hispanic_ethnicity)
print("General Race Categories:", general_race)
```
